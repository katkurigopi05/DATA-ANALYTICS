# Analysis of Call Center Data at Patelco Credit Union
## Example on dealing Unix / POSIX Time

```python
import pandas as pd

# Load the Excel file
df = pd.read_excel("call center data set.xlsx")

# Convert 'Call Start Time' to datetime
df['Call Start Time'] = pd.to_datetime(df['Call Start Time'])

# Extract desired components
df['Weekday'] = df['Call Start Time'].dt.weekday + 1  # Monday=1, Sunday=7
df['DayofMonth'] = df['Call Start Time'].dt.day
df['Hour'] = df['Call Start Time'].dt.hour
df['Minute'] = df['Call Start Time'].dt.minute
df['HM'] = df['Hour'] * 100 + df['Minute']

# Convert 'Call Duration' to timedelta and then to seconds
df['Call Duration'] = pd.to_timedelta(df['Call Duration'])
df['CallDuration-sec'] = df['Call Duration'].dt.total_seconds().astype(int)

# Optional: Select only the columns you need
df_transformed = df[['Agent Name', 'Call Start Time', 'Call Duration','Weekday', 'DayofMonth', 'Hour', 'Minute', 'HM', 'CallDuration-sec']]

# Save the transformed data to a new Excel file
df_transformed.to_excel("Transformed_Call_Center_Data.xlsx", index=False)

print("✅ Transformation complete. Output saved to 'Transformed_Call_Center_Data.xlsx'")
```
